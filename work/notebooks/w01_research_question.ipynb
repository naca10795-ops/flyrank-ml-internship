## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import pandas as pd

# Clone repository if needed
if not os.path.exists('flyrank-ml-internship'):
    !git clone https://github.com/naca10795-ops/flyrank-ml-internship.git

# Load dataset
df = pd.read_csv('flyrank-ml-internship/data/raw/content_refresh_anonymized.csv')

# 1. Total content items evaluated
total_pages = len(df)

# 2. Quick Win / Striking Distance candidates: Pages ranking on Page 2 (Positions 11 to 20) with decent search volume (>50)
striking_distance = df[(df['avg_position'] >= 11) & (df['avg_position'] <= 20) & (df['search_volume'] >= 50)]
candidate_count = len(striking_distance)
candidate_pct = (candidate_count / total_pages) * 100

# 3. Average word count comparison (Striking Distance vs Overall)
avg_words_overall = df['word_count'].mean()
avg_words_candidates = striking_distance['word_count'].mean()

print(f"Total Content Items Analyzed: {total_pages:,}")
print(f"Striking Distance Quick Win Candidates (Positions 11-20, Search Vol >= 50): {candidate_count:,} ({candidate_pct:.2f}% of total)")
print(f"Average Word Count: {avg_words_candidates:,.0f} words (Candidates) vs {avg_words_overall:,.0f} words (Overall)")
```
