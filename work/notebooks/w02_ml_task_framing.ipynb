## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Simple if/else rules (such as if income > 70k and spending > 50 then Target) fail because spending behavior forms non-linear clusters with fuzzy boundaries across age groups and income levels. Hardcoded rules create artificial cutoff boundaries that misclassify customers who sit on the edge of a behavioral threshold. Unsupervised clustering automatically discovers multi-dimensional density patterns and groups customers organically.
