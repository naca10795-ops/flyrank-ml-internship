## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Hardcoded rules create artificial cutoff boundaries that misclassify customers sitting on edge thresholds. Unsupervised clustering automatically discovers multi-dimensional density patterns and groups customers organically.
